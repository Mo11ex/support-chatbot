```python
from datasets import load_dataset
import pandas as pd

# Замени на ID своего датасета
ds = load_dataset("mteb/RuNLUIntentClassification", 'rus')

# Смотрим структуру
print(f"Датасет: mteb/RuNLUIntentClassification")
print(f"Splits: {list(ds.keys())}")

# Берём train
df = pd.DataFrame(ds['train'])

print(f"Строк: {len(df)}")
print(f"Колонки: {list(df.columns)}")

print(f"\n10 примеров:")
print(df.head(10))

# Находим колонку с метками (обычно label, intent, category)
for col in df.columns:
    if df[col].dtype == 'object' or df[col].nunique() < 200:
        print(f"\nКолонка '{col}': {df[col].nunique()} уникальных значений")
        print(df[col].value_counts().head(30).to_string())
```
