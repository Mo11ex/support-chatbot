```python
"""
Ячейка 1: Загрузка эмбеддинг-моделей
"""
from sentence_transformers import SentenceTransformer
import time

EMBEDDERS = {
    "MiniLM-L12": "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2",
    "e5-small":   "intfloat/multilingual-e5-small",
    "mpnet-base": "sentence-transformers/paraphrase-multilingual-mpnet-base-v2",
}

embedder_info = {}

for name, model_id in EMBEDDERS.items():
    print(f"\n📥 Загрузка: {name}...")
    start = time.time()
    model = SentenceTransformer(model_id)
    load_time = time.time() - start
    
    dim = model.get_sentence_embedding_dimension()
    max_len = model.max_seq_length
    
    embedder_info[name] = {
        "dimension": dim,
        "max_seq_length": max_len,
        "load_time": load_time,
    }
    
    print(f"  ✅ Загружена за {load_time:.1f}s")
    print(f"     Размерность эмбеддинга: {dim}")
    print(f"     Макс. длина входа:      {max_len} токенов")
```

```python
"""
Ячейка 2: Создание мини-базы знаний и тест поиска
"""
from sentence_transformers import util
import time

# Имитация чанков базы знаний интернет-магазина
knowledge_base = [
    # Возврат и обмен
    "Возврат товара надлежащего качества возможен в течение 14 дней с момента получения. "
    "Товар должен сохранять товарный вид, потребительские свойства, фабричные ярлыки.",
    
    "Для оформления возврата перейдите в раздел 'Мои заказы', выберите заказ и нажмите "
    "'Оформить возврат'. Заполните форму и выберите причину возврата.",
    
    "Возврат товара ненадлежащего качества (брак) принимается в течение гарантийного срока. "
    "Для возврата бракованного товара обратитесь в службу поддержки с фотографиями дефекта.",
    
    # Доставка
    "Доставка осуществляется курьерской службой СДЭК и Почтой России. "
    "Стоимость доставки рассчитывается при оформлении заказа и зависит от веса и региона.",
    
    "Бесплатная доставка при заказе от 5000 рублей. "
    "Срок доставки: Москва и СПб — 1-2 дня, регионы — 3-7 рабочих дней.",
    
    "Самовывоз доступен из пунктов выдачи СДЭК. Список пунктов и адреса "
    "можно посмотреть на странице оформления заказа.",
    
    # Оплата
    "Оплата возможна банковской картой Visa, MasterCard, МИР, через СБП, "
    "Apple Pay, Google Pay, а также наложенным платежом при получении.",
    
    "Возврат денежных средств осуществляется в течение 10 рабочих дней "
    "на карту, с которой была произведена оплата. При оплате наложенным платежом — "
    "на банковские реквизиты, указанные в заявлении на возврат.",
    
    # Промокоды
    "Промокод вводится на странице оформления заказа в поле 'Промокод' или 'Купон'. "
    "Нажмите 'Применить' для активации скидки. Одновременно можно использовать только один промокод.",
    
    "Актуальные акции и промокоды публикуются в нашем Telegram-канале и в рассылке. "
    "Подпишитесь, чтобы не пропустить скидки до 50%.",
    
    # Гарантия
    "Гарантия на электронику и бытовую технику составляет 12 месяцев. "
    "На одежду, обувь и аксессуары — 30 дней. Гарантийный ремонт осуществляется "
    "через авторизованные сервисные центры.",
    
    # Общее
    "Режим работы интернет-магазина: круглосуточно. Служба поддержки работает "
    "с 9:00 до 21:00 по московскому времени, без выходных.",
    
    "Связаться с нами можно по телефону 8-800-123-45-67 (бесплатно по России), "
    "по email support@example-shop.ru или через чат на сайте.",
]

# Тестовые запросы пользователей
test_queries = [
    "Как вернуть товар?",
    "Сколько стоит доставка в Новосибирск?",
    "Когда вернут деньги за возврат?",
    "Как использовать промокод?",
    "Какая гарантия на наушники?",
    "Во сколько работает поддержка?",
    "Пришёл бракованный товар, что делать?",
    "Можно ли оплатить при получении?",
    "Есть ли самовывоз?",
    "Какой телефон поддержки?",
]

# Тестирование каждой модели
for name, model_id in EMBEDDERS.items():
    print(f"\n{'='*70}")
    print(f"🔍 Тест Retrieval: {name}")
    print(f"{'='*70}")
    
    model = SentenceTransformer(model_id)
    
    # Индексация базы знаний
    start = time.time()
    
    # Для e5 нужен prefix
    if "e5" in name.lower():
        kb_texts = [f"passage: {chunk}" for chunk in knowledge_base]
        query_prefix = "query: "
    else:
        kb_texts = knowledge_base
        query_prefix = ""
    
    kb_embeddings = model.encode(kb_texts, convert_to_tensor=True, show_progress_bar=False)
    index_time = time.time() - start
    print(f"  Индексация {len(knowledge_base)} чанков: {index_time:.2f}s")
    
    # Поиск
    total_search_time = 0
    for query in test_queries:
        query_text = f"{query_prefix}{query}" if query_prefix else query
        
        start = time.time()
        query_emb = model.encode(query_text, convert_to_tensor=True)
        scores = util.cos_sim(query_emb, kb_embeddings)[0]
        search_time = (time.time() - start) * 1000
        total_search_time += search_time
        
        top_k = 3
        top_indices = scores.argsort(descending=True)[:top_k]
        
        print(f"\n  Q: «{query}»  ({search_time:.1f}ms)")
        for rank, idx in enumerate(top_indices):
            chunk_preview = knowledge_base[idx.item()][:80]
            sim = scores[idx.item()].item()
            marker = "✅" if sim >= 0.70 else "⚠️" if sim >= 0.50 else "❌"
            print(f"    {marker} [{rank+1}] sim={sim:.3f} | {chunk_preview}...")
    
    avg_search = total_search_time / len(test_queries)
    print(f"\n  📊 Среднее время поиска: {avg_search:.1f}ms")
```

```python
"""
Ячейка 3: Количественное сравнение моделей
"""
import pandas as pd

# Эталонные ответы (какой чанк ДОЛЖЕН быть Top-1 для каждого запроса)
# Индексы чанков из knowledge_base (0-based)
ground_truth = {
    "Как вернуть товар?": 0,                    # Возврат в течение 14 дней
    "Сколько стоит доставка в Новосибирск?": 4,  # Бесплатная доставка от 5000
    "Когда вернут деньги за возврат?": 7,        # Возврат денег 10 дней
    "Как использовать промокод?": 8,             # Промокод на странице
    "Какая гарантия на наушники?": 10,           # Гарантия на электронику
    "Во сколько работает поддержка?": 11,        # Режим работы
    "Пришёл бракованный товар, что делать?": 2,  # Возврат ненадлежащего качества
    "Можно ли оплатить при получении?": 6,       # Наложенный платёж
    "Есть ли самовывоз?": 5,                     # Пункты выдачи СДЭК
    "Какой телефон поддержки?": 12,              # Контакты
}

comparison = []

for name, model_id in EMBEDDERS.items():
    model = SentenceTransformer(model_id)
    
    if "e5" in name.lower():
        kb_texts = [f"passage: {c}" for c in knowledge_base]
        prefix = "query: "
    else:
        kb_texts = knowledge_base
        prefix = ""
    
    kb_emb = model.encode(kb_texts, convert_to_tensor=True, show_progress_bar=False)
    
    hits_at_1 = 0
    hits_at_3 = 0
    total_sim = 0
    
    for query, expected_idx in ground_truth.items():
        q = f"{prefix}{query}" if prefix else query
        q_emb = model.encode(q, convert_to_tensor=True)
        scores = util.cos_sim(q_emb, kb_emb)[0]
        top_indices = scores.argsort(descending=True)[:3].tolist()
        
        if top_indices[0] == expected_idx:
            hits_at_1 += 1
        if expected_idx in top_indices:
            hits_at_3 += 1
        
        total_sim += scores[expected_idx].item()
    
    n = len(ground_truth)
    comparison.append({
        "Модель": name,
        "Hit@1": f"{hits_at_1}/{n} ({100*hits_at_1/n:.0f}%)",
        "Hit@3": f"{hits_at_3}/{n} ({100*hits_at_3/n:.0f}%)",
        "Avg Similarity": f"{total_sim/n:.3f}",
        "Размерность": embedder_info[name]["dimension"],
    })

df = pd.DataFrame(comparison)
print("\n📊 Сравнение качества Retrieval:")
print(df.to_string(index=False))
df.to_csv("../results/embedder_comparison.csv", index=False)
```
